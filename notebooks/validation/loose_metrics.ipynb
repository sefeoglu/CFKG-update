```python
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score
```

```python
def inverse_analysis(df_ewc, message_type='Inverse'):
    results = []

    for run_id in range(1,6):

        for task_id in range(1,9):
            count = 0
            no_match =0

            task_relations = df_ewc[(df_ewc['run_id']==run_id) & (df_ewc['task_id']==task_id)]
            task_preds_soft = []
            gt_relations = []
            for inverse_index, inverse_row in task_relations.iterrows():
                gt_relation = inverse_row['gt_relation']
                suggestions = inverse_row['suggestions']
                
                if suggestions!="None":
                    candidates = suggestions + [inverse_row['prediction']]
                else:
                    candidates = [inverse_row['prediction']]
                if gt_relation in candidates:
                    task_preds_soft.append(gt_relation)
                else:
                    task_preds_soft.append(inverse_row['prediction'])

                gt_relations.append(gt_relation)
            acc = accuracy_score(gt_relations, task_preds_soft)
            # print(f"Run ID: {run_id}, Task ID: {task_id}, {message_type} Accuracy: {acc}")
            results.append({
                'run_id': run_id,
                'task_id': task_id,
                'accuracy': acc
            })
            
            
    return results
```

```python
df_pred = pd.json_normalize(pd.read_json(validation_files['baseline']).to_dict(orient="records"))
results = inverse_analysis(df_pred, message_type='Canonical')
results_df = pd.DataFrame(results)
results_df.groupby('task_id').mean().T
```
