```python
validation_files={
    'ewc':"/Users/sefika/phd_projects/llm-catastrophic-re/results_all/fewrel_corrected_results/tmlr/ewc/canonical_errors/validation_results.json",
    'mas':"/Users/sefika/phd_projects/llm-catastrophic-re/results_all/fewrel_corrected_results/tmlr/mas/canonical_errors/validation_results.json",
    'baseline':"/Users/sefika/phd_projects/llm-catastrophic-re/results_all/fewrel_corrected_results/tmlr/baseline/canonical_errors/validation_results.json",
    'si':"/Users/sefika/phd_projects/llm-catastrophic-re/results_all/fewrel_corrected_results/tmlr/si/canonical_errors/validation_results.json"   
}
```

```python
def inverse_analysis(df_ewc, message_type='Inverse'):
    results = []
    for run_id in df_ewc['run_id'].unique():
        for task_id in df_ewc['task_id'].unique():
            count = 0
            no_match =0
            inverse = df_ewc[(df_ewc['run_id']==run_id) & (df_ewc['task_id']==task_id) & (df_ewc['message']==message_type)]
            for inverse_index, inverse_row in inverse.iterrows():
                    # print(f"Run ID: {run_id}, Task ID: {task_id}, GT Relation: {gt_relation}, Candidate: {candidate}")
                count += 1  
            # print(f"Run ID: {run_id}, Task ID: {task_id}, Total Inverse Matches: {count}")
            results.append({
                "run_id": run_id,
                "task_id": task_id,
                "total_hallucination": count,
            })
            
    return results
```

```python
df_pred = pd.json_normalize(pd.read_json(validation_files['baseline']).to_dict(orient="records"))
results = inverse_analysis(df_pred, message_type='Hallucinatted_Relation_No_Suggestion')
results_df = pd.DataFrame(results)
canonical_stats = results_df.groupby('task_id').mean(numeric_only=True).reset_index()
canonical_stats.to_json("/Users/sefika/phd_projects/llm-catastrophic-re/results_all/fewrel_corrected_results/tmlr/baseline/canonical_errors/hallucination_stats_baseline.json", orient="records", indent=4)
```
