```python
import matplotlib.pyplot as plt
import pandas as pd
from collections import defaultdict

# Input data
data = [
    {
        "run_id": 1,
        "unique_hallucination": [
            "privileged",
            "blasphemy",
            "country_of_residence",
            "privileged information",
            "state_of_residence"
        ]
    },
    {
        "run_id": 2,
        "unique_hallucination": [
            "opened",
            "countryorprovinces_of_residence",
            "films_attended",
            "country_of_residence",
            "launched",
            "city_of_residence"
        ]
    },
    {
        "run_id": 3,
        "unique_hallucination": [
            "source_of",
            "country_of_employees/members",
            "country"
        ]
    },
    {
        "run_id": 4,
        "unique_hallucination": [
            "religious_affiliation"
        ]
    },
    {
        "run_id": 5,
        "unique_hallucination": []
    }
]
# Count hallucinations per run
run_ids = [entry["run_id"] for entry in data]
hallucination_counts = [len(entry["unique_hallucination"]) for entry in data]

# Plotting bar chart of hallucination counts per run
plt.figure(figsize=(8, 5))
plt.bar(run_ids, hallucination_counts)
plt.xlabel("Run ID")
plt.ylabel("Number of Unique Hallucinations")
plt.title("Unique Hallucinations per Run")
plt.xticks(run_ids)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python


# Collect all hallucinations and count across runs
hallucination_frequency = defaultdict(int)

for entry in data:
    for hallucination in entry["unique_hallucination"]:
        hallucination_frequency[hallucination] += 1

# Convert to DataFrame for plotting
hallucination_df = pd.DataFrame(list(hallucination_frequency.items()), columns=["Hallucination", "Frequency"])
hallucination_df = hallucination_df.sort_values(by="Frequency", ascending=False)

# Plotting with increased font sizes
plt.figure(figsize=(10, 6))
plt.barh(hallucination_df["Hallucination"], hallucination_df["Frequency"])
plt.xlabel("Number of Runs", fontsize=16)
plt.ylabel("Predicted Relations", fontsize=16)
plt.title("Frequency of Hallucinated Relation Types Across Runs", fontsize=18)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.gca().invert_yaxis()
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.savefig('hallucination_frequency_baseline_tacred.pdf', dpi=350)
plt.show()
```
