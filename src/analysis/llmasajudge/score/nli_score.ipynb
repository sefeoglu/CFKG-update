```python

import pandas as pd
import json
path = "/Users/sefika/phd_projects/llm-catastrophic-re/results_all/results_rationale_nli/gemini/tacred_mas_top_prob_5/llm_judged_results.json"
```

```python
def read_json(file_path):
    with open(file_path, "r") as f:
        return json.load(f)

def analyze_tacred_mas(data):
    results = []
    for item in data:
        scores = []
        for record in item['human_judged']:
            try:
                score = record['score'].split('/')[0].split('.')[0]  # Extract the score (e.g., "entailment", "neutral", "contradiction")
                # print(score)
                scores.append(int(score))
            except Exception as e:
                print(f"Error processing record: {record}, error: {e}")   

        results.append({
            'run_id': item['run_id'],
            'scores': scores,
            'len_scores': len(scores)
        })
    return results
```

```python
tacred_mas = analyze_tacred_mas(read_json(path))
tacred_mas_df = pd.DataFrame(tacred_mas)
tacred_mas_df
```
