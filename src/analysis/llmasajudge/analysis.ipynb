```python

import pandas as pd
import json
path = "/Users/sefika/phd_projects/llm-catastrophic-re/results_rationale/gemini/tacred_mas_top_prob_5/llm_judged_results.json"
```

```python
def read_json(file_path):
    with open(file_path, "r") as f:
        return json.load(f)

def analyze_tacred_mas(data):
    llm_judged = [item['llm_judged'] for item in data]
    return llm_judged
```

```python
tacred_mas = analyze_tacred_mas(read_json(path))
tacred_mas_df = pd.DataFrame(tacred_mas)
tacred_mas_df.T.head()
```

```python
fewrel_path = "/Users/sefika/phd_projects/llm-catastrophic-re/results_rationale/gemini/fewrel_mas_top_prob_5/llm_judged_results.json"

fewrel = analyze_tacred_mas(read_json(fewrel_path))
fewrel_df = pd.DataFrame(fewrel)
fewrel_df.T.head()
```
