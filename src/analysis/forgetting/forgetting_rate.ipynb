```python
import os
import json
from sklearn.metrics import multilabel_confusion_matrix, accuracy_score
import numpy as np
import pandas as pd

def read_json(path):
    with open(path, 'r', encoding="utf-8") as f:
        data = json.load(f)
    return data

def write_json(data, path):
    if not os.path.exists(os.path.dirname(path)):
        os.makedirs(os.path.dirname(path))
    with open(path, 'w', encoding="utf-8") as f:
        json.dump(data, f, indent=4, ensure_ascii=False)
```

```python
import numpy as np

def forgetting_rate(R):
    """
    Calculate average forgetting rate in continual learning.

    Parameters:
        R (2D list or np.ndarray): Matrix of accuracies where R[i][j] is
                                   performance on task j after training on task i
                                   (0-indexed). Shape: (T, T)

    Returns:
        float: Average forgetting rate.
        list: Forgetting for each task.
    """
    R = np.array(R)
    T = R.shape[0]
    forgettings = []

    for j in range(T - 1):  # last task can't be forgotten
        max_acc = np.max(R[j:, j])  # best performance after training it
        final_acc = R[T - 1, j]     # final performance after all tasks
        forgettings.append(max_acc - final_acc)

    avg_forgetting = np.mean(forgettings)
    std_forgetting = np.std(forgettings)
    return avg_forgetting, forgettings, std_forgetting

```

```python
data1 = read_json(experiments['run_1'])
data2 = read_json(experiments['run_2'])
data3 = read_json(experiments['run_3'])
data4 = read_json(experiments['run_4'])
data5 = read_json(experiments['run_5'])
```

```python
data1 = pd.DataFrame(data1)
data2 = pd.DataFrame(data2)
data3 = pd.DataFrame(data3)
data4 = pd.DataFrame(data4)
data5 = pd.DataFrame(data5)
```

```python
combined = pd.concat([data1, data2, data3, data4, data5])

# Compute average accuracy
average_accuracy = (
    combined
    .groupby(['base_task', 'task'], as_index=False)
    .agg({'accuracy': 'mean'})
)

# Optional: Pivot into matrix form
accuracy_matrix = average_accuracy.pivot(index='base_task', columns='task', values='accuracy')
```

```python

avg_F, per_task_F, std_F = forgetting_rate(accuracy_matrix)
print(f"Average Forgetting Rate: {avg_F*100:.2f}%")
print(f"Per Task Forgetting Rates: {per_task_F}")
print(f"Standard Deviation of Forgetting Rates: {std_F}")
```
