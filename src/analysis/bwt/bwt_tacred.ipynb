```python
import os
import json
from sklearn.metrics import multilabel_confusion_matrix, accuracy_score
import numpy as np
import pandas as pd

def read_json(path):
    with open(path, 'r', encoding="utf-8") as f:
        data = json.load(f)
    return data

def write_json(data, path):
    if not os.path.exists(os.path.dirname(path)):
        os.makedirs(os.path.dirname(path))
    with open(path, 'w', encoding="utf-8") as f:
        json.dump(data, f, indent=4, ensure_ascii=False)
```

```python
def bwt(forgetting_matrix):
    num_tasks = 8
    bwt_values = []
    for item in forgetting_matrix:

        for base in range(1, num_tasks):
            for task in range(1, num_tasks+1):
                if item['base_task']== base and item['task']== task:
                    base_acc= item['accuracy']
                if item['base_task']== task and item['task']== base:
                    result_acc= item['accuracy']
        bwt_task = result_acc - base_acc
      
        bwt_values.append(bwt_task)
    overall_bwt = np.mean(bwt_values)
    return overall_bwt
```

```python
def compute_avg_bwt(forgetting_matrix_paths):
    bwt_results = {}
    mean_bwt_values = []
    for i, path in forgetting_matrix_paths.items():
        forgetting_matrix = np.array(read_json(path))
   
        overall_bwt = bwt(forgetting_matrix)
        mean_bwt_values.append(overall_bwt)
        bwt_results[path] = {
            "overall_bwt": overall_bwt
        }
    bwt_results['mean_bwt'] = np.mean(mean_bwt_values)*100
    bwt_results['std_bwt'] = np.std(mean_bwt_values)*100
    return bwt_results
```
